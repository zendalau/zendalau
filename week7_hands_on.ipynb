## DISPLAY PANDA DATAFRAMES - selecting rows & columns ##

```python
import pandas as pd
```

```python
print('Instead of using print, we can also use "DISPLAY"')
```

```python
display("DataFrame_name")
```

```python
fruits=pd.read_csv("fruits.csv")
display(fruits)
```

```python
print("to select columns, create a list")
fruits[['Country','Fruits']]
```

```python
print("to select only ONE column, put the variable name as a string")
fruits["Prices"]
```

```python
print("to select rows, use iloc. If multiple rows, the indices must be put in a list first. If ONE row, the index can be put in the index operator directly. Row indexes start with 0 and ends with -1")
```

```python
display(fruits)
print("Select row 0, aka 1st row")
(fruits.iloc[0])
```

```python
print("Select all rows until BEFORE -1 aka the 2nd last row")
(fruits.iloc[:-1])
```

```python
print("Select row 0 to BEFORE row 2")
(fruits.iloc[0:2])
```

```python
print("Select row 0 and row 2, and column 0 and column 2")
fruits.iloc[[0,2],[0,2]]
```

```python
print("Select 1st row and 3rd row, keep all columns.")
fruits.iloc[[0,2], :]
```

```python
print('Select row 0 to BEFORE row 2 from fruits and prices')
fruits[['Fruits','Prices']].iloc[0:2]
```

## BOOLEAN MASKING - using conditions to identify the data selection ##

```python
display(fruits)
```

```python
print('Select those with prices > 0.7')
fruits[ (fruits['Prices'] > 0.7) ]
```

```python
print('Select those elements which satisfy BOTH conditions using "&". EACH CONDITION MUST BE ENCLOSED IN PARENTHESES ( )')
fruits[ (fruits['Prices'] > 0.7) & (fruits['Country'] == 'New Zealand')]
```

```python
print('Select those elements which satisfy EITHER conditions using "|". EACH CONDITION MUST BE ENCLOSED IN PARENTHESES ( )')
fruits[ (fruits['Prices'] > 0.7) | (fruits['Country'] == 'New Zealand')]
```

## MERGING DATAFRAME - by rows ##

```python
print('Create data 1 using a dictionary')
data1 = {'Customer ID':[1,2],'Product':['TV', 'Oven']}
df1 = pd.DataFrame(data1)
display (df1)
```

```python
print('Create data 2 using a dictionary')
data2 = {'Customer ID':[3,4],'Product':['Oven','TV']}
df2 = pd.DataFrame(data2)
display (df2)
```

```python
print('CONCAT data 1 and data 2 together. NOTE THE INDEX FOR THE ROWS')
df3 = pd.concat([df1,df2])
display (df3)
```

```python
print('CONCAT data 1 and data 2 together, BUT keep the index for the rows in a consecutive manner. Type "ignore_index = True"')
df4 = pd.concat([df1,df2], ignore_index=True)
display (df4)
```

## MERGING DATAFRAME - natural join: return the rows where 2 dataframes have matching keys ##

```python
data5 = {'Customer ID':[1,2,3,4,5],'Product':['TV','TV','TV','Oven','Oven']}
df5 = pd.DataFrame(data5)
display (df5)

data6 = {'Customer ID':[2,4,6,7],'Shop':['Shop_A','Shop_A','Shop_B','Shop_C']}
df6 = pd.DataFrame(data6)
display (df6)
```

```python
print('Select data using INNER JOIN based on Customer ID. Only Customer ID 2 and 4 showed up in both dataframes.')
inner_df = pd.merge(df5, df6, on='Customer ID', how='inner')
display (inner_df)
```

## MISSING DATA & OUTLIERS - find missing values ##

```python
c_imports = pd.read_csv('fruits_corrupted.csv')
print('Notice how Apple prices is NaN = Not a Number')
c_imports
```

```python
print('To check every cell in the dataframe and returns TRUE if the cell contains a missing value. TRUE = Missing value FALSE = Value exist')
c_imports.isnull()
```

```python
print('is.null returns TRUE for missing values. .sum adds up all the TRUE , counting the number of missing values')
print('axis = 0 will count the columns . The results below shows "Prices" has 1 TRUE aka missing value')
c_imports.isnull().sum(axis=0)
```

```python
print('is.null returns TRUE for missing values. .sum adds up all the TRUE , counting the number of missing values')
print('axis = 1 will count the rows. The results below show row 0 has 1 TRUE aka missing value')
c_imports.isnull().sum(axis=1)
```

```python
print('is.null().any(axis __) is used to check whether ANY missing data exist')
c_imports.isnull().any()
```

```python
print('Create a variable called missing_cimport to represent the missing values.')
missing_cimport = c_imports.isnull().any()
print('use == True.index to specifically identify WHICH index from the columns has missing values.')
missing_cimport [missing_cimport == True].index
```

## MISSING DATA & OUTLIERS - replace missing values ##

```python
price_mean = c_imports['Prices'].mean()
```

```python
print('Replace missing values by using .fillna()')
c_imports['Prices'].fillna(value=price_mean)
```

## MISSING DATA & OUTLIERS - deleting missing data ##

```python
print('.drop is used to specify the ROW INDICES with missing values that should be removed.')
c_imports.drop(axis=0, index = [0])
```

```python
print('.dropna is to drop the missing values. axis = 0(rows) 1(columns). how = "all" means to remove if ALL VALUES ARE MISSING')
print('results show to drop the columns if ALL values in the column is missing. NO COLUMN IS HAS ALL MISSING VALUES')
c_imports.dropna(axis=1, how="all")
```

```python
print('how = "any" means to remove if ANY VALUES ARE MISSING. ROW 0 IS REMOVED BECAUSE THERE IS ONE MISSING VALUE')
c_imports.dropna(axis=0, how="any")
```
