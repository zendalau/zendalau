```python
print("LOGICAL OPERATORS")
print("True if both operators are true. and")
a=7
b=11
print(a!=b and a==7)

print("True if either one is true. or")
a=7
b=11
print(a==b or a==11)

print("Used to reverse the logical state of the expression -> True become False, False become True. not")
a=7
b=11
print(a==b)
print(not a==b)
```
