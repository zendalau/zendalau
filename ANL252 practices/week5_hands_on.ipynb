```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data_url = "https://raw.githubusercontent.com/selva86/datasets/refs/heads/master/BostonHousing.csv"
```

```python
print('This line assigns a URL string to a variable. To actually load the data, use pandas')
```

```python
raw_data = np.genfromtxt(data_url, delimiter=",", skip_header=1)

print(raw_data)
```

```python
print('Loads the CSV file from the URL into a NumPy array delimiter="," → CSV format. Skip_header=1 → skips the column names (first row)')
```

```python
data = raw_data[:, :-1]

print(data)
```

```python
print('Selects all rows. Selects all columns except the last one. These are the feature variables (inputs)')
```

```python
target = raw_data[:, -1]

print(target)
```

```python
print('Selects all rows. Selects the last column only. This is the target variable (output, e.g. house price)')
```

```python
housing_data = data

print(housing_data)
```

```python
print('Assigns the feature data to a new variable name. housing_data and data now reference the same array')
```
