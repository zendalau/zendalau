```python
import pandas as pd

# Display the raw data values
df = pd.read_csv("MissingValue.csv")
df.set_index('ID',inplace=True)
display(df)
```

```python
# Report the missing values
display(df.isnull())
display(df[df.isnull().any(axis=1)])
```

```python
# Describe the statistical summaries
df.describe(include='all')
```

```python
import pandas as pd

# Detect the outliers
df_outlier = pd.read_csv('MissingValue_Outlier.csv')
df_outlier.set_index('ID', inplace=True)
display(df_outlier)

Q1 = df_outlier.quantile(0.25)
Q3 = df_outlier.quantile(0.75)
IQR = Q3-Q1
lower_bound = Q1-(1.5*IQR)
upper_bound = Q3+(1.5*IQR)
outliers = (df_outlier < lower_bound)|(df_outlier > upper_bound)

display(df_outlier[outliers.any(axis=1)])
print('This outlier is ID 10')

```
